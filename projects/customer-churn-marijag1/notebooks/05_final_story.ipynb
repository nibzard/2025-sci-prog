# 05 – Završni zaključak i rasprava

U ovoj bilježnici sažimam cijeli projekt: cilj, korištene podatke, glavne uvide iz analize te rezultate modeliranja.
Na kraju navodim zaključak (koji model je bolji) i preporuke što bi se moglo poboljšati.


## Sažetak projekta

Projekt je obuhvaćao sljedeće korake:
1. Učitavanje i opis podataka
2. Čišćenje i pripremu podataka
3. Eksploratornu analizu podataka (EDA)
4. Izgradnju i usporedbu modela strojnog učenja
5. Evaluaciju performansi modela

Cilj je bio predvidjeti hoće li korisnik napustiti uslugu na temelju dostupnih atributa.


## Rezultati

Analiza podataka pokazala je da oko **26.5% korisnika napušta uslugu**, dok oko **73.5% ostaje**, što ukazuje na umjerenu neuravnoteženost klasa.

Najvažniji faktori koji utječu na churn:
- trajanje korištenja usluge (Tenure Months)
- vrsta ugovora (Contract)
- mjesečni trošak (Monthly Charges)
- vrsta internetske usluge
- dostupnost tehničke podrške

Korisnici s kratkim ugovorima (Month-to-month) i višim mjesečnim troškovima imaju veću vjerojatnost napuštanja usluge.


## Modeli

Izgrađena su dva modela:
- Logistic Regression
- Random Forest Classifier

Modeli su evaluirani pomoću metrika: accuracy, precision, recall, F1-score i ROC-AUC.

Logistic Regression je pokazao viši ROC-AUC i recall za churn korisnike, dok je Random Forest postigao veću ukupnu točnost.
S obzirom da je u ovom problemu važnije identificirati korisnike koji će napustiti uslugu, Logistic Regression je odabran kao preferirani model.


## Rasprava

Rezultati pokazuju da financijski i ugovorni faktori imaju najveći utjecaj na churn, dok demografski podaci imaju manji utjecaj.
Ovo je u skladu s poslovnom logikom, jer korisnici češće napuštaju uslugu zbog cijene ili kvalitete usluge, a ne zbog osobnih karakteristika.

Modeli su sposobni predvidjeti churn s relativno visokom točnošću, što pokazuje potencijal za primjenu u realnom poslovnom okruženju.


## Ograničenja

Projekt ima nekoliko ograničenja:
- Dataset predstavlja podatke samo jedne telekom kompanije
- Nema vremenske serije ponašanja korisnika
- Klase su neuravnotežene, što može utjecati na performanse modela

## Budući rad

U budućem radu moguće je:
- koristiti tehnike balansiranja podataka (SMOTE)
- isprobati naprednije modele poput XGBoost ili Neural Networks
- izraditi web aplikaciju za predikciju churn-a
- koristiti SHAP za objašnjenje modela (explainable AI)


## Zaključak

Projekt je uspješno demonstrirao primjenu metoda znanstvenog programiranja i strojnog učenja na stvarnom poslovnom problemu.
Predloženi modeli mogu pomoći telekom kompanijama u identifikaciji korisnika s visokim rizikom odlaska i omogućiti proaktivne strategije zadržavanja korisnika.
